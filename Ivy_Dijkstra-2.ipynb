```python
from math import inf
import heapq

def dijkstra(graph, source, destination):
    #Initialize distances (all infinity except source)
    dist = {node: inf for node in graph}
    dist[source] = 0

    # Store the path (to reconstruct later)
    parent = {node: None for node in graph}

    # Priority queue (min-heap) to get smallest distance node
    pq = [(0, source)]  # (distance, node)

    while pq:
        # Get node with smallest distance
        current_dist, current_node = heapq.heappop(pq)

        # Skip if we already found a better path
        if current_dist > dist[current_node]:
            continue

        # Stop early if we reached destination
        if current_node == destination:
            break

        # Check all neighbors
        for neighbor, weight in graph[current_node]:
            new_distance = current_dist + weight

            # If shorter path found, update it
            if new_distance < dist[neighbor]:
                dist[neighbor] = new_distance
                parent[neighbor] = current_node
                heapq.heappush(pq, (new_distance, neighbor))

    #If destination is unreachable
    if dist[destination] == inf:
        return inf, []

    #Reconstruct shortest path
    path = []
    current = destination

    while current:
        path.append(current)
        current = parent[current]

    path.reverse()  # Reverse to get correct order

    return dist[destination], path


def print_result(distance, path):
    #Print results
    if distance == inf:
        print("Shortest distance: No path exists")
        print("Shortest path: None")
    else:
        print("Shortest distance:", distance)
        print("Shortest path:", " -> ".join(path))


# Graph with locations around Montclair
# I use real locations around MSU as nodes to make the graph more relatable
graph = {
    "Montclair": [("Upper Montclair", 4), ("Bloomfield", 7), ("MSU Campus", 2)],
    "Upper Montclair": [("Montclair", 4), ("Bloomfield", 3), ("Little Falls", 6)],
    "Bloomfield": [("Montclair", 7), ("Upper Montclair", 3), ("Clifton", 4)],
    "Clifton": [("Bloomfield", 4), ("MSU Campus", 5)],
    "MSU Campus": [("Montclair", 2), ("Clifton", 5), ("Little Falls", 3)],
    "Little Falls": [("MSU Campus", 3), ("Upper Montclair", 6)]
}

# Choose start and destination
source = "MSU Campus"
destination = "Bloomfield"

# Step 11: Run Dijkstra
distance, path = dijkstra(graph, source, destination)

# Step 12: Display result
print_result(distance, path)
```
